# Joints on the wire: the SO-101 server node, its controllers, and the arm client

The camera pattern, applied to the thing that can actually pinch you.
**Exactly one process owns the Feetech serial bus**: the server node. It is
the hardware side of **ros2_control**: the controllers sit between it and
every client, and turn a move into a smooth trajectory.

![Alt Text](../../animations/SO-101_Loop_Animation.gif)

**This notebook moves the real arm.** The server below connects straight
to the SO-101's serial bus, so every command in this notebook turns into
physical motion. Clear the space around the arm and keep hands away from
the joints.

The conventions that make this safe (and your TODOs implement):

- behaviors speak **degrees**, keyed `shoulder_pan, shoulder_lift,
  elbow_flex, wrist_flex, wrist_roll, gripper`; ROS speaks **radians**, and
  the joint map converts at the server's and the client's topics;
  **partial commands are legal**;
- every command passes `clamp_joints`, on the client *and* the server
  (belt and braces);
- the state's `effort` carries the **gripper's torque in N·m**, from its
  servo current, with **NaN meaning "not readable"**. That feedback lets
  the grip behavior stop squeezing at first contact instead of crushing
  whatever it holds.

## 1. A minimal arm server: ros2_control's hardware side

The demo twin of TODOs 2.4-2.6, with one difference from the camera
notebook: the backend is the real thing. `make_arm_backend(...)` connects
a `LeRobotArm` to the Feetech serial bus using the `robot:` section of the
workshop config, and this notebook becomes the ONE process that owns the
bus. If the arm isn't reachable, the cell tells you what to check instead
of running.

The server never talks to behaviors. It talks to `topic_based_ros2_control`,
the controllers' hardware plugin: setpoints arrive on
`/so101/hardware/joint_commands` and the arm's state goes back on
`/so101/hardware/joint_states`, both `sensor_msgs/JointState` in
**radians**. The joint map (`so101_sim.mapping.JointMap`) converts to and
from the bus's degrees. The state also carries each joint's velocity, from
the change in position since the last tick, and the gripper's torque in
N·m (NaN when the bus can't read it).

Note the **control thread**: subscriptions land on executor threads, but
only one sequential loop touches the arm. That loop is the only thing on
the serial bus. No contention, ever. It writes every fresh setpoint as it
comes - the controllers interpolate the motion - through LeRobot's per-tick
clamp, the transport's own backstop.

````python
require_ros()

from rclpy.node import Node
from rclpy.qos import HistoryPolicy, QoSProfile, ReliabilityPolicy
from sensor_msgs.msg import JointState

SENSOR_QOS = QoSProfile(
    reliability=ReliabilityPolicy.BEST_EFFORT, history=HistoryPolicy.KEEP_LAST, depth=1
)
sim = load_sim_config(cfg)
jmap = JointMap(sim["joint_map"], sim["gripper_map"])  # radians on the wire <-> degrees on the bus


def open_real_arm():
    """Connect to the real SO-101 over the Feetech serial bus, streaming:
    every setpoint is one write, through LeRobot's per-tick clamp."""
    try:
        return make_arm_backend(server_transport_cfg(cfg))  # LeRobotArm per the robot: config
    except Exception as e:
        raise RuntimeError(
            "could not connect to the SO-101. To fix:\n"
            f"  - check the USB cable and that {cfg['robot']['motor_port']} exists\n"
            "  - check port permissions (sudo usermod -aG dialout $USER, then log out/in)\n"
            "  - make sure nothing else owns the bus (a running arm_node, another kernel)\n"
            "  - if this arm is new to this machine, calibrate once in a terminal first\n"
            f"then re-run this cell. (underlying error: {e})"
        ) from e


class DemoArmServer(Node):
    def __init__(self, control_fps=60.0):
        # The client finds the server's parameters under this name (the real
        # server's gripper_torque_limit_nm lives there).
        super().__init__(SERVER_NODE)
        self.arm = open_real_arm()
        self._lock = threading.Lock()
        self._target = dict(REST_POSE)
        self._fresh_until = 0.0
        self.create_subscription(JointState, HARDWARE_COMMANDS, self.on_command, SENSOR_QOS)
        self.state_pub = self.create_publisher(JointState, HARDWARE_STATES, SENSOR_QOS)
        self._running = True
        self._loop = threading.Thread(target=self.control_loop, args=(control_fps,), daemon=True)
        self._loop.start()

    def on_command(self, msg):
        joints = {
            n: jmap.to_cal(n, float(p)) for n, p in zip(msg.name, msg.position) if n in MOTOR_NAMES
        }  # radians -> degrees; partial dicts OK
        if joints:
            with self._lock:
                self._target.update(clamp_joints(joints))  # safety backstop
                self._fresh_until = time.time() + 1.0

    def control_loop(self, fps):
        # Every fresh setpoint is written: the controllers interpolate the
        # move, and the transport's per-tick clamp is the backstop. When
        # setpoints stop, the arm holds its last pose.
        last = None
        while self._running:
            with self._lock:
                target, fresh = dict(self._target), time.time() < self._fresh_until
            if fresh:
                self.arm.send_joints(target)
            joints = self.arm.get_joints()
            now = time.monotonic()
            load = self.arm.get_gripper_load()  # Present_Current counts; None if not readable
            rad = [jmap.to_rad(m, joints[m]) for m in MOTOR_NAMES]
            velocity = None if last is None else [(a - b) / (now - last[0]) for a, b in zip(rad, last[1])]
            last = (now, rad)
            self.state_pub.publish(
                joint_state_msg(
                    MOTOR_NAMES,
                    rad,
                    velocity=velocity,
                    effort=[
                        math.nan if (load is None or m != "gripper")
                        else current_to_torque_nm(load, sim["current_model"])
                        for m in MOTOR_NAMES
                    ],
                    node=self,
                )
            )
            time.sleep(1.0 / fps)

    def close(self):
        self._running = False
        self._loop.join()
        self.arm.close()  # parks at REST_POSE and releases the bus
        self.destroy_node()


server = DemoArmServer()
host.add_node(server)
print("arm server up - sole owner of the real Feetech bus, on", HARDWARE_COMMANDS, "and", HARDWARE_STATES)
````

## 2. The controllers in between

`launch/so101_control.launch.py` starts ros2_control on top of the server:

- **`topic_based_ros2_control`**, the hardware plugin: it publishes the
  controllers' setpoints to the server and reads the server's state back
  (section 1's two topics). It only publishes while the arm is away from
  the setpoint, so a still arm costs no traffic;
- **`arm_controller`** (a `JointTrajectoryController`): a move is one
  `FollowJointTrajectory` goal - where to go, and how long the servos'
  profile says it takes. The controller interpolates it, then reports
  whether the joints got there within the goal's tolerance;
- **`arm_forward_controller`** and **`gripper_forward_controller`**:
  setpoints passed straight through, for streamed motion (gesture mimic,
  person tracking) and for the gripper;
- **`joint_state_broadcaster`**: the arm's state for everyone, on
  `/joint_states`.

The controllers run at `robot.control_fps`, the server's rate. Their
launch shuts down with this kernel.

## 3. The client side, and the first rule of robot clients

**Never command an arm whose state you can't see.** The client blocks until
`/joint_states` and the controllers are live, *then* exposes the `ArmClient`
API. That part, and running a move as one trajectory goal, comes from the
provided base, `common.ros2_control_arm.ControllerArm`. A client wires the
endpoints and the two conversions - exactly TODOs 2.7-2.9:

- the endpoints: the trajectory action client, the forward controllers'
  publishers, and the `/joint_states` subscription;
- a streamed setpoint: clamped, then sent in radians;
- a state: radians back to degrees, and the gripper's effort with NaN as
  `None`.

````python
require_ros()

from control_msgs.action import FollowJointTrajectory
from rclpy.action import ActionClient
from std_msgs.msg import Float64MultiArray


class DemoArmClient(ControllerArm):
    def _setup_ros(self, node):
        self._trajectory = ActionClient(node, FollowJointTrajectory, f"/{ARM_CONTROLLER}/follow_joint_trajectory")
        self._stream_pub = node.create_publisher(Float64MultiArray, f"/{STREAM_CONTROLLER}/commands", 10)
        self._gripper_pub = node.create_publisher(Float64MultiArray, f"/{GRIPPER_CONTROLLER}/commands", 10)
        node.create_subscription(JointState, "/joint_states", self._on_state, 10)

    def _on_state(self, msg):
        positions, effort = parse_joint_state(msg)  # radians; NaN->None handled here
        self._update_state(
            {n: self._jmap.to_cal(n, p) for n, p in positions.items() if n in MOTOR_NAMES},
            joint_velocity(msg, "gripper"),
            effort,
        )

    def send_joints(self, joints):
        joints = clamp_joints(joints, self._send_limits)  # client-side clamp too
        if "gripper" in joints:
            self._command_gripper(joints["gripper"])
        body = self._stream_setpoint(joints)  # the whole body setpoint, or None
        if body is not None:
            self._stream_pub.publish(Float64MultiArray(data=[self._jmap.to_rad(k, body[k]) for k in BODY]))


arm = DemoArmClient(cfg, node_name="demo_arm_client")
print(
    "state stream and controllers live. rest pose:",
    {k: round(v, 1) for k, v in arm.get_joints().items()},
)
print(
    "gripper effort:",
    arm.get_gripper_load(),
    " (Present_Current counts, from the N·m on the wire; None means not readable, NOT zero)",
)
````

## 4. Drive every joint by hand

Before scripting motion, get a feel for the real arm. One slider per
joint, ranges straight from `JOINT_LIMITS`. Drag a slider and **release**
it: the release value becomes one move - one trajectory goal, timed by the
servos' own profile - and the controller drives the arm along it. One
gesture, one goal. (Why on release? A running cell only receives a drag's
flood of widget syncs throttled, while one sync per gesture always lands.)
A move runs to its end; a gesture released meanwhile waits, and only the
newest waiting one runs.

The sliders start at `REST_POSE` (wrist_flex rests at 22.5 on this rig)
and the cell sends the arm there first, so the UI and the metal agree from
the first pixel. The gripper slider spans the jaws' calibrated range,
43 (closed) to 95 (open).

Two extras:

- **link shoulder_lift + elbow_flex** (on by default): the two proximal
  joints move together, in opposition - move shoulder_lift by +0.1 and
  elbow_flex follows by -0.1, in the SAME goal - which keeps the forearm
  attitude roughly steady as the shoulder moves. Untick to drive them
  independently.
- **rest pose** sends ONE full-pose goal, then quietly syncs the sliders
  to match.

The loop below just keeps UI events flowing and streams the *measured*
joint state back under the sliders. Press **stop** to end the loop and
move to the next cell.

These sliders move the physical arm. Move them gently and keep clear.

````python
require_ros()

LINKED = ("shoulder_lift", "elbow_flex")  # move together, in opposition

# One worker runs the moves, one at a time; the UI only queues them. The
# queue holds one: a newer gesture replaces an older one that has not started.
goals = queue.Queue(maxsize=1)
DONE = object()


def submit(goal):
    try:
        goals.get_nowait()  # drop the older gesture that has not started
    except queue.Empty:
        pass
    goals.put_nowait(goal)


def mover():
    while (goal := goals.get()) is not DONE:
        arm.move_to(goal)


worker = threading.Thread(target=mover, daemon=True)
worker.start()
submit(dict(REST_POSE))  # start from a known pose: rest - the sliders below default to it too

sliders = {}
for name in MOTOR_NAMES:
    lo, hi = JOINT_LIMITS[name]
    sliders[name] = widgets.FloatSlider(
        value=REST_POSE[name],
        min=lo,
        max=hi,
        step=0.1 if name in LINKED else 0.5,
        description=name,
        continuous_update=False,  # one state sync per gesture, on release
        readout_format=".1f",
        layout=widgets.Layout(width="440px"),
        style={"description_width": "110px"},
    )

link_chk = widgets.Checkbox(
    value=True,
    indent=False,
    description="link shoulder_lift + elbow_flex (opposite)",
    layout=widgets.Layout(width="340px"),
)

# Event-driven goals: each slider CHANGE queues one partial goal; the
# controller holds every joint it leaves out. With the link on, a
# shoulder_lift move of +d takes elbow_flex along by -d (and vice versa) in
# the SAME goal, so the pair stays coordinated.
syncing = {"on": False}  # guards programmatic slider writes


def make_sender(name):
    def _send(change):
        if syncing["on"]:
            return
        cmd = {name: float(change["new"])}
        if link_chk.value and name in LINKED:
            other = LINKED[1] if name == LINKED[0] else LINKED[0]
            delta = float(change["new"]) - float(change["old"])
            partner = sliders[other]
            target = max(partner.min, min(partner.max, partner.value - delta))
            syncing["on"] = True
            try:
                partner.value = target  # mirror the UI without re-sending
            finally:
                syncing["on"] = False
            cmd[other] = float(target)
        submit(cmd)  # ONE goal for the whole gesture

    return _send


senders = {name: make_sender(name) for name in sliders}
for name, s in sliders.items():
    s.observe(senders[name], "value")

stop_btn = widgets.Button(description="stop", button_style="danger")
rest_btn = widgets.Button(description="rest pose", button_style="warning")
stopped = {"flag": False}
stop_btn.on_click(lambda _: stopped.update(flag=True))


def go_rest(_):
    submit(dict(REST_POSE))  # ONE full-pose goal...
    syncing["on"] = True  # ...then sync the UI quietly
    try:
        for n in MOTOR_NAMES:
            sliders[n].value = REST_POSE[n]
    finally:
        syncing["on"] = False


rest_btn.on_click(go_rest)

readout = widgets.Label()
display(
    widgets.VBox(
        [*sliders.values(), link_chk, widgets.HBox([rest_btn, stop_btn]), readout]
    )
)

# The loop's only jobs: deliver the widget events (poll) and stream the
# MEASURED state back so you can watch the arm track each goal.
with ui_events() as poll:
    while not stopped["flag"]:
        poll(10)  # slider releases + clicks land here
        j = arm.get_joints()
        readout.value = "state:  " + "   ".join(f"{n}={j[n]:.1f}" for n in MOTOR_NAMES)
        time.sleep(0.03)

for name, s in sliders.items():  # detach: the sliders go quiet
    s.unobserve(senders[name], "value")  # once you leave this cell
submit(DONE)  # the worker finishes the move it runs, then ends
worker.join()
print("joint control loop stopped - sliders detached")
````

## 5. The safety clamp survives the transport

`clamp_joints` runs on both ends. Command something insane and watch physics
stay bounded. Heads up: the arm really will swing to its +86 degree pan
limit and back (the software clamp - the mechanical range pulled in 25% so
the arm clears the environment), so keep the space around it open anyway.
`move_to` returns when the controller reports the move over: `"arrived"`
(within the goal's tolerance), `"blocked"` (it ended short) or `"stopped"`.

````python
require_ros()

pan_lo, pan_hi = JOINT_LIMITS["shoulder_pan"]
how = arm.move_to({"shoulder_pan": 500.0})  # 500 degrees into an +/-86 degree clamp
pan = arm.get_joints()["shoulder_pan"]
print(f"commanded 500 -> the move {how} at {pan:.1f}  (limit: {pan_lo:+.0f}/{pan_hi:+.0f})")
arm.move_to({"shoulder_pan": 0.0})
````

## 6. Why the effort field exists

The gripper servo reports its **Present_Current** through the bus. The
server turns it into torque (N·m, along the STS3215 datasheet line) and
sends it in `JointState.effort`, sampled every few control ticks; the
client turns it back into current counts, so the grip behavior keeps its
thresholds. Watching current instead of commanded position is what lets a
grip behavior stop closing at first sustained contact rather than driving
to a target and crushing the object on the way.

Two conventions make it work over any transport, and you implement both:
NaN on the wire means "not readable" and the client maps it to `None`,
never zero; when effort is unavailable, behaviors fall back to detecting a
position stall instead.
